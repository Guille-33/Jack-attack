````python
import sqlite3
import pandas as pd

connection = sqlite3.connect("data/sql-murder-mystery.db")
cursor= connection.cursor()

def get_df(query:str)->pd.DataFrame:
    pd.set_option('display.max_colwidth', None)
    resp=cursor.execute(query)
    col = [d[0] for d in cursor.description]
    return(pd.DataFrame(resp, columns = col))
````

````python
query='''
SELECT type, city, description
FROM crime_scene_report
WHERE date=20180115 and type="murder" and city="SQL City"'''
get_df(query=query)
````

````python
query='''
SELECT id , name, address_number
FROM person
WHERE address_street_name= "Northwestern Dr"
ORDER BY address_number DESC
'''
get_df(query)
````

````python
query='''
SELECT id , name , address_street_name
FROM person
WHERE name LIKE "%Annabell%"
'''
get_df(query)
````

````python
query='''
select transcript
from interview
where person_id=14887
'''
get_df(query)
````

````python
query='''
select F.id, F.person_id, F.name, P.license_id, D.plate_number
from get_fit_now_member as F
left JOIN person as P
on F.person_id = P.id
left join drivers_license as D
on P.license_id=D.id
where F.id like "48Z%" and F.membership_status="gold" and D.plate_number like "%H42W%"
'''
get_df(query)
````
