```python
import random
from faker import Faker
from config import (N_CLIENTES,N_PEDIDOS,N_PRODUCTOS)
from BQ_dataset import upload_table,create_client
import pandas as pd
MAPA_PAISES = {
    "España": Faker('es_ES'),
    "Alemania": Faker('de_DE'),
    "Italia": Faker('it_IT'),
    "Francia": Faker('fr_FR')
}
```

```python
estados_producto=['activo','inactivo']
estados_producto_list=[]
for i in range(len(estados_producto)):
    estados_producto_list.append({
        'id':i+1,
        'estado':estados_producto[i]
    })
df_estados_producto=pd.DataFrame(estados_producto_list)
```

```python
fake = Faker('es_ES')

categorias_list = []
costes_list=[]
productos_list=[]

# Rango de precios coherente
categorias_tech = {
    "Portátiles y PCs": {
        "items": [
            "Laptop", "Ordenador de Sobremesa", "Mini PC", "Estación de Trabajo", 
            "All-in-One", "Servidor Doméstico", "Laptop Convertible 2-en-1"
        ],
        "precio_rango": (450.0, 2500.0)
    },
    "Smartphones y Tablets": {
        "items": [
            "Teléfono Inteligente", "Tablet", "E-Reader", "Smartphone Plegable", 
            "Reloj Inteligente Premium", "Phablet"
        ],
        "precio_rango": (150.0, 1200.0)
    },
    "Audio y Sonido": {
        "items": [
            "Auriculares Inalámbricos", "Altavoz Inteligente", "Barra de Sonido", 
            "Micrófono USB", "Auriculares de Estudio", "Amplificador Digital",
            "Subwoofer Inalámbrico", "Sistema Home Cinema"
        ],
        "precio_rango": (30.0, 400.0)
    },
    "Componentes y Periféricos": {
        "items": [
            "Teclado Mecánico", "Ratón Gaming", "Monitor 4K", "Tarjeta Gráfica", 
            "Disco SSD", "Memoria RAM DDR5", "Placa Base", "Refrigeración Líquida",
            "Hub USB-C Multipuerto", "Alfombrilla RGB"
        ],
        "precio_rango": (20.0, 800.0)
    }
}

# Elementos para construir nombres comerciales realistas
marcas_ficticias = [
    "Nexus", "Apex", "Quantum", "Volt", "Zeta", "Titan", "Aura", "Krypton",
    "Aero", "Cortex", "Synapse", "Astra", "Vector", "Hydra", "Nova", "Pulse",
    "Helix", "Chronos", "Specter", "Matrix", "Zenith", "Vortex"
]
Sufijos = [
    "Pro", "Max", "Ultra", "X", "Lite", "Plus", "Air", "Gaming Edition",
    "Elite", "Master", "Prime", "Studio", "Carbon", "Ultimate", "Neo", "Evo"
]

# Conjunto para asegurar que ningún nombre completo se repita
nombres_unicos = set()

for i in range(1, N_PRODUCTOS + 1):
    tipo = random.choice(list(categorias_tech.keys()))
    descripcion = random.choice(list(categorias_tech[tipo].get('items')))
    precio_min,precio_max=categorias_tech[tipo].get('precio_rango')
    precio_venta=round(random.uniform(precio_min,precio_max),2)
    # 2. Construir un nombre
    while True:
        marca = random.choice(marcas_ficticias)
        sufijo = random.choice(Sufijos)
        nombre_completo = f"{marca} {sufijo}"
        
        if nombre_completo not in nombres_unicos:
            nombres_unicos.add(nombre_completo)
            break
            
    
    categorias_list.append({
        "id": i,
        "tipo":tipo,
        "nombre": nombre_completo,
        "descripcion":descripcion
    })
    costes_list.append({
        "id":i,
        "precio_venta":precio_venta,
        "coste":round(random.uniform(precio_min,precio_venta),2)
    })
    productos_list.append({
        "id":i,
        "categoria_id":i,
        "coste_id":i,
        "stock":random.randint(5,100),
        "estado":random.randint(1,len(df_estados_producto))
    })

# Cargar en el DataFrame de Pandas
df_categorias = pd.DataFrame(categorias_list)
df_costes=pd.DataFrame(costes_list)
df_productos=pd.DataFrame(productos_list)
```

```python
canales=['organic','paid ads','redes sociales','empresa']
canales_list=[]
for i in range(len(canales)):
    canales_list.append({
        'id':i+1,
        'canal':canales[i]
    })
df_canales=pd.DataFrame(canales_list)
```

```python
estados_pedido=['pending','confirmed','shipped','delivered','cancelled','returned']
estados_pedido_list=[]
for i in range(len(estados_pedido)):
    estados_pedido_list.append({
        'id':i+1,
        'estado':estados_pedido[i]
    })
df_estados_pedido=pd.DataFrame(estados_pedido_list)
```

```python
estados_pago=['completado','reembolsado','pendiente','fallido']
estados_pago_list=[]
for i in range(len(estados_pago)):
    estados_pago_list.append({
        'id':i+1,
        'estado':estados_pago[i]
    })
df_estados_pago=pd.DataFrame(estados_pago_list)
```

```python
clientes_list=[]
for i in range(1, N_CLIENTES + 1):
    pais=random.choice(list(MAPA_PAISES.keys()))
    fake= MAPA_PAISES[pais]
    clientes_list.append({
        "id_cliente": i,
        "nombre": fake.first_name(),
        "apellido":fake.last_name(),
        "telefono": fake.unique.numerify(text='6########'),  # Formato móvil único
        "pais":pais,
        "ciudad": fake.city(),
        "canal":random.randint(1,len(df_canales)),
        "fecha_registro": fake.date_between(start_date='-7y', end_date='-5y')
    })
df_clientes = pd.DataFrame(clientes_list)
```

```python
plantillas_reviews = {
    5: ["¡Excelente compra! ", "Cumple perfectamente con lo prometido. ", "Espectacular, superó mis expectativas. ", "Calidad brutal. "],
    4: ["Muy buen producto, llegó a tiempo. ", "Funciona correctamente y se nota de buena calidad. ", "Relación calidad-precio inbatible. "],
    3: ["Está bien, pero tardó un poco en llegar. ", "Funciona, aunque los materiales podrían ser mejores. ", "Normal, correcto para el precio. "],
    2: ["No me ha convencido del todo. ", "Tiene fallos intermitentes. ", "La descripción no se ajusta al 100% a la realidad. "],
    1: ["Pésimo producto, vino defectuoso. ", "Una decepción absoluta, no lo recomiendo. ", "No funciona, tramitando la devolución. ", "Malísima calidad. "]
}

pedidos_list=[]
timeline_list=[]
lista_pedidos_list=[]
valoraciones_list = []
pagos_list=[]

for i in range(1,N_PEDIDOS+1):
    # pedido no puede ser antes de fecha de creacion de usuario, para no comprobar he decidido dejar un año de separacion entre creacion y pedido
    f_pedido=fake.date_between(start_date='-4y', end_date='-1y')
    f_entrega=fake.date_between(start_date=f_pedido,)
    f_envio=fake.date_between(start_date=f_pedido,end_date=f_entrega) #envio debe estar entre entrega y pedido
    cliente_id=random.randint(1,len(df_clientes))
    timeline_list.append({
        'id':i,
        'pedido': f_pedido,
        'envio': f_envio,
        'entrega':f_entrega
    })
    pedidos_list.append({
        'id':i,
        'cliente_id':cliente_id,
        'estado_pedido_id':random.choice([4,6]), # como estos son datos pasados solo pueden estar o entregados o devueltos
        'timeline_id':i
    })
    total=0
    for j in range (random.randint(1,4)):
        cantidad=random.randint(1,5)
        compra=round(random.uniform(50.0,2500.0),2)
        lista_pedidos_list.append({
            'id':i*10+j,
            'pedido_id':i,
            'producto_id':random.randint(1,len(df_productos)),
            'cantidad':cantidad,
            'precio_compra':compra,
            'descuento':round(random.uniform(0,1),2)
        })
        if random.randint(1,3)==1:
            rating=random.randint(1,5)
            valoraciones_list.append({
                'id':len(valoraciones_list)+1,
                'cliente_id':cliente_id,
                'linea_pedido_id': i*10+j,
                'rating':rating,
                'comentario': random.choice(plantillas_reviews[rating])
            })
        total+=compra*cantidad
    pagos_list.append({
        'id':i,
        'pedido_id':i,
        'importe':total,
        'fecha':f_pedido,
        'estado_pago_id':1 # como son datos anteriores asumimos que estan todos pagados
    })
df_timeline=pd.DataFrame(timeline_list)
df_pedidos=pd.DataFrame(pedidos_list)
df_pagos=pd.DataFrame(pagos_list)
df_valoraciones=pd.DataFrame(valoraciones_list)
df_lista_pedidos=pd.DataFrame(lista_pedidos_list)
```
