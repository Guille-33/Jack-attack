```python
from BQ_dataset import (create_client,generate_dataset,build_BQ_table)

client=create_client()
```

```python
generate_dataset(client=client)
```

```python
canal={
    'id':['INT64','REQUIRED'],
    'canal':['STRING','REQUIRED']
}
categoria={
    'id':['INT64','REQUIRED'],
    'tipo':['STRING','REQUIRED'],
    'nombre':['STRING','REQUIRED'],
    'descripcion':['STRING','REQUIRED']
}
costes={
    'id':['INT64','REQUIRED'],
    'precio_venta':['FLOAT64','REQUIRED'],
    'coste':['FLOAT64','REQUIRED']
}
estado_productos={
    'id':['INT64','REQUIRED'],
    'estado':['STRING','REQUIRED']
}
registro={
    'id':['INT64','REQUIRED'],
    'pedido':['DATE','REQUIRED'],
    'envio':['DATE'],
    'entrega':['DATE']
}
estado_pedido={
    'id':['INT64','REQUIRED'],
    'estado':['STRING','REQUIRED']
}
estado_pago={
    'id':['INT64','REQUIRED'],
    'estado':['STRING','REQUIRED']
}
constraints={
    'primary':'id'
}
build_BQ_table(name='channels',fields=canal,constraints=constraints,client=client)
build_BQ_table(name='categories',fields=categoria,constraints=constraints,client=client)
build_BQ_table(name='costs',fields=costes,constraints=constraints,client=client)
build_BQ_table(name='product_status',fields=estado_productos,constraints=constraints,client=client)
build_BQ_table(name='timeline',fields=registro,constraints=constraints,client=client)
build_BQ_table(name='order_status',fields=estado_pedido,constraints=constraints,client=client)
build_BQ_table(name='payment_status',fields=estado_pago,constraints=constraints,client=client)
```

```python
customer={
    'id':['INT64','REQUIRED'],
    'nombre':['STRING','REQUIRED'],
    'apellido':['STRING','REQUIRED'],
    'telefono':['INT64','REQUIRED'],
    'pais':['STRING','REQUIRED'],
    'ciudad':['STRING','REQUIRED'],
    'canal_id':['INT64','REQUIRED'],
    'fecha':['DATE','REQUIRED'],
}
constraints={
    'primary':'id',
    'foreigns':[{
        'ref_table':'channels',
        'ref_column':'id',
        'loc_column':'canal_id'
    }]
}
build_BQ_table(name='customers',fields=customer,constraints=constraints,client=client)
```

```python
productos={
    'id':['INT64','REQUIRED'],
    'categoria_id':['INT64','REQUIRED'],
    'coste_id':['INT64','REQUIRED'],
    'stock':['INT64','REQUIRED'],
    'estado_producto_id':['INT64','REQUIRED'],
}
constraints={
    'primary':'id',
    'foreigns':[
        {
            'ref_table':'categories',
            'ref_column':'id',
            'loc_column':'categoria_id'
        },
        {
            'ref_table':'costs',
            'ref_column':'id',
            'loc_column':'coste_id'
        },
        {
            'ref_table':'product_status',
            'ref_column':'id',
            'loc_column':'estado_producto_id'
        },
    ],
}
build_BQ_table(name='products',fields=productos,constraints=constraints,client=client)
```

```python
pedidos={
    'id':['INT64','REQUIRED'],
    'cliente_id':['INT64','REQUIRED'],
    'estado_pedido_id':['INT64','REQUIRED'],
    'timeline_id':['INT64','REQUIRED'],
}
constraints={
    'primary':'id',
    'foreigns':[
        {
            'ref_table':'customers',
            'ref_column':'id',
            'loc_column':'cliente_id'
        },
        {
            'ref_table':'order_status',
            'ref_column':'id',
            'loc_column':'estado_pedido_id'
        },
        {
            'ref_table':'timeline',
            'ref_column':'id',
            'loc_column':'timeline_id'
        },
    ],
}
build_BQ_table(name='orders',fields=pedidos,constraints=constraints,client=client)
```

```python
lineas_pedido={
    'id':['INT64','REQUIRED'],
    'pedido_id':['INT64','REQUIRED'],
    'producto_id':['INT64','REQUIRED'],
    'cantidad':['INT64','REQUIRED'],
    'precio_compra':['FLOAT64','REQUIRED'],
    'descuento':['FLOAT64']
}
constraints={
    'primary':'id',
    'foreigns':[
        {
            'ref_table':'orders',
            'ref_column':'id',
            'loc_column':'pedido_id'
        },
        {
            'ref_table':'products',
            'ref_column':'id',
            'loc_column':'producto_id'
        },
    ]
}
build_BQ_table(name='order_items',fields=lineas_pedido,constraints=constraints,client=client)
```

```python
pagos={
    'id':['INT64','REQUIRED'],
    'pedido_id':['INT64','REQUIRED'],
    'importe':['INT64','REQUIRED'],
    'fecha':['DATE','REQUIRED'],
    'estado_pago_id':['INT64','REQUIRED'],
}
constraints={
    'primary':'id',
    'foreigns':[
        {
            'ref_table':'orders',
            'ref_column':'id',
            'loc_column':'pedido_id'
        },
        {
            'ref_table':'payment_status',
            'ref_column':'id',
            'loc_column':'estado_pago_id'
        },
    ]
}
build_BQ_table(name='payments',fields=pagos,constraints=constraints,client=client)
```

```python
reseñas={
    'id':['INT64','REQUIRED'],
    'cliente_id':['INT64','REQUIRED'],
    'linea_pedido_id':['INT64','REQUIRED'],
    'rating':['INT64','REQUIRED'],
    'comentario':['STRING'],
}
constraints={
    'primary':'id',
    'foreigns':[
        {
            'ref_table':'customers',
            'ref_column':'id',
            'loc_column':'cliente_id'
        },
        {
            'ref_table':'order_items',
            'ref_column':'id',
            'loc_column':'linea_pedido_id'
        },
    ]
}
build_BQ_table(name='reviews',fields=reseñas,constraints=constraints,client=client)
```
